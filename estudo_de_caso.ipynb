# PARA O MACHINE LEARNING

```python
albardao = albardao[
    [
        "uf",
        "nome_uc",
        "data_inicio",
        "mediana",
        "media",
        "minimo",
        "maximo",
    ]
]
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Garante que os dados estejam ordenados cronologicamente (janela fixa de 8 dias)
albardao = albardao.copy()
albardao["data_dt"] = pd.to_datetime(albardao["data_inicio"])
albardao = albardao.sort_values("data_dt").reset_index(drop=True)

# 2. Define t como o tempo decorrido em dias desde a primeira data da série (vetor 1D)
primeira_data = albardao["data_dt"].iloc[0]
t = (albardao["data_dt"] - primeira_data).dt.days.to_numpy()

# 3. Define y como a variável alvo (média da clorofila-a)
y = albardao["media"].to_numpy()

# 4. Define o período sazonal anual (365,25 dias) e a frequência angular omega (2*pi / período)
periodo = 365.25
w = 2 * np.pi / periodo

# 5. Monta a matriz de variáveis preditoras: tendência linear (t) + 1º, 2º e 3º harmônicos anuais
X_sazonal = np.column_stack(
    [
        t,  # Tendência linear ao longo do tempo
        np.cos(w * t),
        np.sin(w * t),  # 1º harmônico (ciclo de 1 ano)
        np.cos(2 * w * t),
        np.sin(2 * w * t),  # 2º harmônico (meio ano)
        np.cos(3 * w * t),
        np.sin(3 * w * t),  # 3º harmônico (1/3 de ano)
    ]
)

# 6. Ajusta o modelo linear por mínimos quadrados aos dados históricos
modelo_sazonal = LinearRegression()
modelo_sazonal.fit(X_sazonal, y)

# 7. Calcula os valores ajustados dentro do período histórico
previsao_historica = modelo_sazonal.predict(X_sazonal)

# 8. Gera as próximas 23 datas futuras em passos fixos de 8 dias (~6 meses após a última data)
ultima_data = albardao["data_dt"].iloc[-1]
datas_futuras = pd.date_range(
    start=ultima_data + pd.Timedelta(days=8),
    periods=40,
    freq="8D",
)

# 9. Converte as datas futuras para a mesma escala t (dias desde a primeira data histórica)
t_futuro = (datas_futuras - primeira_data).days.to_numpy()

# 10. Monta a matriz X_futuro com exatamente as mesmas colunas usadas no treino
X_futuro = np.column_stack(
    [
        t_futuro,
        np.cos(w * t_futuro),
        np.sin(w * t_futuro),
        np.cos(2 * w * t_futuro),
        np.sin(2 * w * t_futuro),
        np.cos(3 * w * t_futuro),
        np.sin(3 * w * t_futuro),
    ]
)

# 11. Calcula a previsão para as janelas futuras de 8 dias
previsao_futura = modelo_sazonal.predict(X_futuro)

# 12. Calcula o desvio padrão dos resíduos para montar uma faixa aproximada de incerteza
residuos = y - previsao_historica
sigma_res = np.std(residuos, ddof=X_sazonal.shape[1] + 1)
ic_superior = previsao_futura + 1.96 * sigma_res
ic_inferior = np.maximum(0, previsao_futura - 1.96 * sigma_res)

# 13. Plota histórico, ajuste, previsão futura e faixa aproximada
fig, ax = plt.subplots(figsize=(12, 6))
ax.scatter(
    albardao["data_dt"],
    y,
    alpha=0.45,
    s=22,
    color="#1f77b4",
    label="Observado (janela de 8 dias)",
)
ax.plot(
    albardao["data_dt"],
    previsao_historica,
    color="#2ca02c",
    linewidth=2,
    label="Ajuste histórico (harmônicos 1º–3º)",
)
ax.plot(
    datas_futuras,
    previsao_futura,
    color="#d62728",
    linewidth=2.2,
    linestyle="--",
    label="Previsão futura (passos de 8 dias)",
)
ax.fill_between(
    datas_futuras,
    ic_inferior,
    ic_superior,
    color="#d62728",
    alpha=0.16,
    label="Faixa aproximada (±1,96 DP residual)",
)
ax.axvline(ultima_data, color="#555555", linestyle=":", linewidth=1.2)
ax.set_xlabel("Data de Início (Janela de 8 dias)", fontsize=11)
ax.set_ylabel("Média Clorofila-a ($mg/m^3$)", fontsize=11)
ax.legend(loc="upper left")
plt.tight_layout()
plt.savefig("previsão.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Garante que os dados estejam ordenados cronologicamente (janela fixa de 8 dias)
albardao = albardao.copy()
albardao["data_dt"] = pd.to_datetime(albardao["data_inicio"])
albardao = albardao.sort_values("data_dt").reset_index(drop=True)

# 2. Separa treino (até 2025) e teste (2026)
treino_df = albardao[albardao["data_dt"].dt.year <= 2025].copy()
teste_df = albardao[albardao["data_dt"].dt.year == 2026].copy()

# Usa a primeira data do treino como origem t = 0
primeira_data = treino_df["data_dt"].iloc[0]
t_treino = (treino_df["data_dt"] - primeira_data).dt.days.to_numpy()
y_treino = treino_df["media"].to_numpy()

t_teste = (teste_df["data_dt"] - primeira_data).dt.days.to_numpy()
y_teste = teste_df["media"].to_numpy()

# 3. Define o período sazonal anual (365,25 dias) e a frequência angular omega
periodo = 365.25
w = 2 * np.pi / periodo


def montar_features(t_vetor, k_max=3):
    # Monta tendência linear (t) + pares (cos, sin) do 1º até o k_max-ésimo harmônico
    colunas = [t_vetor]
    for k in range(1, k_max + 1):
        colunas.append(np.cos(k * w * t_vetor))
        colunas.append(np.sin(k * w * t_vetor))
    return np.column_stack(colunas)


X_treino = montar_features(t_treino)
X_teste = montar_features(t_teste)

# 4. Ajusta o modelo exclusivamente nos dados de treino (até 2025)
modelo_sazonal = LinearRegression()
modelo_sazonal.fit(X_treino, y_treino)

# 5. Avalia no treino e no teste (2026)
ajuste_treino = modelo_sazonal.predict(X_treino)
previsao_2026 = modelo_sazonal.predict(X_teste)

if len(y_teste) > 0:
    mae_2026 = mean_absolute_error(y_teste, previsao_2026)
    rmse_2026 = np.sqrt(mean_squared_error(y_teste, previsao_2026))
    r2_2026 = r2_score(y_teste, previsao_2026) if len(y_teste) > 1 else np.nan
    print(f"Teste 2026 (n={len(y_teste)}) -> MAE: {mae_2026:.4f} | RMSE: {rmse_2026:.4f} | R2: {r2_2026:.4f}")

# 6. Gera datas futuras em passos de 8 dias para 2027
# 6. Gera datas futuras em passos de 8 dias a partir do último ponto disponível de 2026
ultima_data_2026 = (
    teste_df["data_dt"].iloc[-1]
    if len(teste_df) > 0
    else albardao["data_dt"].iloc[-1]
)
inicio_previsao = ultima_data_2026 + pd.Timedelta(days=8)
fim_2027 = pd.Timestamp("2027-12-31")
datas_2027 = pd.date_range(start=inicio_previsao, end=fim_2027, freq="8D")

t_2027 = (datas_2027 - primeira_data).days.to_numpy()
X_2027 = montar_features(t_2027)
previsao_2027 = modelo_sazonal.predict(X_2027)

# 7. Calcula faixa aproximada de incerteza a partir dos resíduos do treino
residuos_treino = y_treino - ajuste_treino
sigma_res = np.std(residuos_treino, ddof=X_treino.shape[1] + 1)

ic_sup_2026 = previsao_2026 + 1.96 * sigma_res
ic_inf_2026 = np.maximum(0, previsao_2026 - 1.96 * sigma_res)

ic_sup_2027 = previsao_2027 + 1.96 * sigma_res
ic_inf_2027 = np.maximum(0, previsao_2027 - 1.96 * sigma_res)

# 8. Plota treino (até 2025), teste (2026) e previsão futura (2027)
fig, ax = plt.subplots(figsize=(13, 6))

ax.scatter(
    treino_df["data_dt"],
    y_treino,
    alpha=0.45,
    s=20,
    color="#1f77b4",
    label="Observado Treino (≤ 2025)",
)
ax.scatter(
    teste_df["data_dt"],
    y_teste,
    alpha=0.75,
    s=26,
    color="#17becf",
    label="Observado Teste (2026)",
)
ax.plot(
    treino_df["data_dt"],
    ajuste_treino,
    color="#2ca02c",
    linewidth=2,
    label="Ajuste Treino (≤ 2025)",
)
ax.plot(
    teste_df["data_dt"],
    previsao_2026,
    color="#ff7f0e",
    linewidth=2.2,
    linestyle="-.",
    label="Avaliação Teste (2026)",
)
ax.fill_between(
    teste_df["data_dt"],
    ic_inf_2026,
    ic_sup_2026,
    color="#ff7f0e",
    alpha=0.12,
)
ax.plot(
    datas_2027,
    previsao_2027,
    color="#d62728",
    linewidth=2.2,
    linestyle="--",
    label="Previsão Futura (2027)",
)
ax.fill_between(
    datas_2027,
    ic_inf_2027,
    ic_sup_2027,
    color="#d62728",
    alpha=0.16,
    label="Faixa aproximada (±1,96 DP residual)",
)

ax.axvline(pd.Timestamp("2026-01-01"), color="#555555", linestyle=":", linewidth=1.2)
ax.axvline(pd.Timestamp("2027-01-01"), color="#555555", linestyle="--", linewidth=1.2)
ax.set_xlabel("Data de Início (Janela de 8 dias)", fontsize=11)
ax.set_ylabel("Média Clorofila-a ($mg/m^3$)", fontsize=11)
ax.legend(loc="upper left")
plt.tight_layout()
plt.savefig("previsão_2.png", dpi=300)
plt.show()
```
